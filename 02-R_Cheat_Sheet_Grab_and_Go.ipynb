## 4. Tipuri de variabile și conversii

`character` stochează text; `factor` reprezintă categorii. O variabilă categorială poate fi importată inițial ca `character`.

```r
# Verifică tipul
class(DATASET$VARIABILA)
str(DATASET)

# Convertește și salvează rezultatul
DATASET$VARIABILA <- as.numeric(DATASET$VARIABILA)
DATASET$VARIABILA <- as.character(DATASET$VARIABILA)
DATASET$VARIABILA <- as.factor(DATASET$VARIABILA)
```

```r
# Variabilă categorială ordinală
DATASET$VARIABILA <- factor(
  DATASET$VARIABILA,
  levels = c("NIVEL_MIC", "NIVEL_MEDIU", "NIVEL_MARE"),
  ordered = TRUE
)
```

```r
# Text cu unități → număr
DATASET$VARIABILA <- parse_number(DATASET$VARIABILA)

# Factor numeric → număr, fără codurile interne ale factorului
DATASET$VARIABILA <- as.numeric(as.character(DATASET$VARIABILA))
```

## 5. Selectare, filtrare și variabile noi

```r
# Păstrează anumite coloane
DATASET_MIC <- DATASET |>
  select(VARIABILA_1, VARIABILA_2, VARIABILA_3)
```

```r
# Păstrează rândurile care respectă condiția
DATASET_FILTRAT <- DATASET |>
  filter(VARIABILA_NUMERICA > PRAG)

# & = și; | = sau
DATASET_FILTRAT <- DATASET |>
  filter(VARIABILA_1 > PRAG_1 & VARIABILA_2 == "CATEGORIE")
```

```r
# Creează sau modifică variabile
DATASET <- DATASET |>
  mutate(
    VARIABILA_NOUA = FORMULA,
    VARIABILA_BINARA = if_else(CONDITIE, "Da", "Nu")
  )
```

```r
# Text → dată și oră
DATASET$VARIABILA_TIMP <- as.POSIXct(
  DATASET$VARIABILA_TIMP,
  format = "%Y-%m-%d %H:%M:%S"
)
```

## 7. Statistici descriptive

```r
# Centru
mean(DATASET$VARIABILA, na.rm = TRUE)
median(DATASET$VARIABILA, na.rm = TRUE)

# Răspândire
var(DATASET$VARIABILA, na.rm = TRUE)
sd(DATASET$VARIABILA, na.rm = TRUE)
min(DATASET$VARIABILA, na.rm = TRUE)
max(DATASET$VARIABILA, na.rm = TRUE)
range(DATASET$VARIABILA, na.rm = TRUE)
```

```r
# Frecvențe și proporții
FRECVENTE <- table(DATASET$VARIABILA)
FRECVENTE
prop.table(FRECVENTE)

# Modul sau modurile statistice
MODURI <- names(FRECVENTE[FRECVENTE == max(FRECVENTE)])
MODURI

# Pentru o variabilă numerică:
as.numeric(MODURI)

# Atenție: mode() NU calculează modul statistic.
```

```r
# Rezumat pe grupuri
REZUMAT <- DATASET |>
  group_by(VARIABILA_GRUP) |>
  summarise(
    n = n(),
    medie = mean(VARIABILA_NUMERICA, na.rm = TRUE),
    mediana = median(VARIABILA_NUMERICA, na.rm = TRUE),
    varianta = var(VARIABILA_NUMERICA, na.rm = TRUE),
    deviatia_standard = sd(VARIABILA_NUMERICA, na.rm = TRUE)
  )

REZUMAT
```

## 8. Grafice în `ggplot2`

| Întrebare | Funcție principală |
|---|---|
| Evoluție în timp | `geom_line()` |

### 8.1. Evoluție în timp

```r
ggplot(DATASET, aes(x = VARIABILA_TIMP, y = VARIABILA_NUMERICA)) +
  geom_line(color = "NUME_CULOARE") +
  labs(
    title = "TITLU_GRAFIC",
    x = "ETICHETA_AXA_X",
    y = "ETICHETA_AXA_Y"
  ) +
  theme_minimal()
```

```r
ggplot(DATASET, aes(x = VARIABILA_NUMERICA)) +
  geom_histogram(
    bins = NUMAR_BINURI,
    fill = "NUME_CULOARE",
    color = "white"
  ) +
  labs(
    title = "TITLU_GRAFIC",
    x = "ETICHETA_AXA_X",
    y = "Frecvență"
  ) +
  theme_minimal()
```

### 8.3. Scatterplot

```r
ggplot(DATASET, aes(x = VARIABILA_X, y = VARIABILA_Y)) +
  geom_point(alpha = 0.7, color = "NUME_CULOARE") +
  labs(
    title = "TITLU_GRAFIC",
    x = "ETICHETA_AXA_X",
    y = "ETICHETA_AXA_Y"
  ) +
  theme_minimal()
```

```r
# Adaugă o linie de tendință liniară
ggplot(DATASET, aes(x = VARIABILA_X, y = VARIABILA_Y)) +
  geom_point(alpha = 0.7) +
  geom_smooth(method = "lm", se = FALSE) +
  theme_minimal()
```

## 10. Covarianță și corelație

```r
cov(
  DATASET$VARIABILA_X,
  DATASET$VARIABILA_Y,
  use = "complete.obs"
)

cor(
  DATASET$VARIABILA_X,
  DATASET$VARIABILA_Y,
  use = "complete.obs"
)
```

## 11. Rețetă completă de analiză

```r
library(readr)
library(dplyr)
library(ggplot2)

# 1. Import și verificare
DATASET <- read_csv("NUME_FISIER.csv")
head(DATASET)
str(DATASET)
summary(DATASET)

# 2. Curățare
DATE_ANALIZA <- DATASET |>
  filter(!is.na(VARIABILA_X), !is.na(VARIABILA_Y))

# 3. Centru și răspândire
mean(DATE_ANALIZA$VARIABILA_X, na.rm = TRUE)
median(DATE_ANALIZA$VARIABILA_X, na.rm = TRUE)
var(DATE_ANALIZA$VARIABILA_X, na.rm = TRUE)
sd(DATE_ANALIZA$VARIABILA_X, na.rm = TRUE)

# 4. Distribuție
ggplot(DATE_ANALIZA, aes(x = VARIABILA_X)) +
  geom_histogram(bins = NUMAR_BINURI) +
  labs(title = "TITLU", x = "AXA X", y = "Frecvență") +
  theme_minimal()

# 5. Relație
ggplot(DATE_ANALIZA, aes(x = VARIABILA_X, y = VARIABILA_Y)) +
  geom_point() +
  labs(title = "TITLU", x = "AXA X", y = "AXA Y") +
  theme_minimal()

cor(
  DATE_ANALIZA$VARIABILA_X,
  DATE_ANALIZA$VARIABILA_Y,
  use = "complete.obs"
)

# 6. Export
write_csv(DATE_ANALIZA, "NUME_FISIER_NOU.csv")
ggsave("NUME_GRAFIC.png", width = 8, height = 5, dpi = 300)
```
