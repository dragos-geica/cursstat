## 1. Setup

```r
library(readr)
library(dplyr)
library(ggplot2)
```

## 2. Importarea datelor

```r
DATASET <- read_csv("NUME_FISIER.csv")
```

```r
# Dacă separatorul este ;
DATASET <- read_csv2("NUME_FISIER.csv")
```

## 3. Verificarea rapidă a datelor

```r
head(DATASET)
tail(DATASET)
str(DATASET)
summary(DATASET)
nrow(DATASET)
ncol(DATASET)
names(DATASET)
```

## 4. Selectarea și filtrarea observațiilor

```r
# Păstrează anumite coloane
DATASET_MIC <- DATASET |>
  select(VARIABILA_1, VARIABILA_2, VARIABILA_3)
```

```r
# Păstrează observațiile care respectă o condiție
DATASET_FILTRAT <- DATASET |>
  filter(VARIABILA_NUMERICA > PRAG)
```

```r
# Mai multe condiții: & = și, | = sau
DATASET_FILTRAT <- DATASET |>
  filter(VARIABILA_1 > PRAG_1 & VARIABILA_2 == "CATEGORIE")
```

```r
# Exclude explicit valorile lipsă
DATASET_FILTRAT <- DATASET |>
  filter(!is.na(VARIABILA), VARIABILA > PRAG)
```

## 5. Crearea și modificarea variabilelor

```r
DATASET <- DATASET |>
  mutate(
    VARIABILA_NOUA = FORMULA,
    VARIABILA_BINARA = if_else(CONDITIE, "Da", "Nu")
  )
```

```r
# Transformă textul în categorie
DATASET <- DATASET |>
  mutate(VARIABILA_CATEGORICA = factor(VARIABILA_CATEGORICA))
```

```r
# Transformă timestamp-ul în dată și oră
DATASET <- DATASET |>
  mutate(timestamp = as.POSIXct(timestamp, format = "%Y-%m-%d %H:%M:%S"))
```

## 6. Valorile lipsă (`NA`)

```r
is.na(DATASET$VARIABILA)
sum(is.na(DATASET$VARIABILA))
mean(is.na(DATASET$VARIABILA)) * 100
```

```r
# Calcule ignorând NA
mean(DATASET$VARIABILA, na.rm = TRUE)
median(DATASET$VARIABILA, na.rm = TRUE)
sd(DATASET$VARIABILA, na.rm = TRUE)
```

```r
# Păstrează doar observațiile complete pentru variabilele alese
DATASET_COMPLET <- DATASET |>
  filter(!is.na(VARIABILA_1), !is.na(VARIABILA_2))
```

## 7. Statistici descriptive

```r
mean(DATASET$VARIABILA, na.rm = TRUE)
median(DATASET$VARIABILA, na.rm = TRUE)
var(DATASET$VARIABILA, na.rm = TRUE)
sd(DATASET$VARIABILA, na.rm = TRUE)
min(DATASET$VARIABILA, na.rm = TRUE)
max(DATASET$VARIABILA, na.rm = TRUE)
range(DATASET$VARIABILA, na.rm = TRUE)
```

```r
# Modul pentru o variabilă categorială sau discretă
table(DATASET$VARIABILA)
names(which.max(table(DATASET$VARIABILA)))
```

```r
# Rezumat pe grupuri
REZUMAT <- DATASET |>
  group_by(VARIABILA_GRUP) |>
  summarise(
    n = n(),
    medie = mean(VARIABILA_NUMERICA, na.rm = TRUE),
    mediana = median(VARIABILA_NUMERICA, na.rm = TRUE),
    deviatia_standard = sd(VARIABILA_NUMERICA, na.rm = TRUE)
  )

REZUMAT
```

## 8. Alegerea graficului

| Întrebare | Grafic |
|---|---|
| Evoluție în timp | Line graph |
| Distribuția unei variabile numerice | Histogram |
| Relația dintre două variabile numerice | Scatterplot |
| Comparație între categorii | Bar chart |

### 8.1. Line graph

```r
ggplot(DATASET, aes(x = VARIABILA_TIMP, y = VARIABILA_NUMERICA)) +
  geom_line(color = "NUME_CULOARE") +
  labs(
    title = "TITLU_GRAFIC",
    x = "ETICHETA_AXA_X",
    y = "ETICHETA_AXA_Y"
  ) +
  theme_minimal()
```

### 8.2. Histogram

```r
ggplot(DATASET, aes(x = VARIABILA_NUMERICA)) +
  geom_histogram(bins = NUMAR_INTERVALe, fill = "NUME_CULOARE", color = "white") +
  labs(
    title = "TITLU_GRAFIC",
    x = "ETICHETA_AXA_X",
    y = "Frecvență"
  ) +
  theme_minimal()
```

### 8.3. Scatterplot

```r
ggplot(DATASET, aes(x = VARIABILA_X, y = VARIABILA_Y)) +
  geom_point(alpha = 0.7, color = "NUME_CULOARE") +
  labs(
    title = "TITLU_GRAFIC",
    x = "ETICHETA_AXA_X",
    y = "ETICHETA_AXA_Y"
  ) +
  theme_minimal()
```

### 8.4. Bar chart

```r
# R numără automat observațiile din fiecare categorie
ggplot(DATASET, aes(x = VARIABILA_CATEGORICA)) +
  geom_bar(fill = "NUME_CULOARE") +
  labs(
    title = "TITLU_GRAFIC",
    x = "CATEGORIE",
    y = "Număr de observații"
  ) +
  theme_minimal()
```

```r
# Compară o statistică deja calculată pe categorii
ggplot(REZUMAT, aes(x = VARIABILA_GRUP, y = medie)) +
  geom_col(fill = "NUME_CULOARE") +
  labs(
    title = "TITLU_GRAFIC",
    x = "CATEGORIE",
    y = "MEDIE"
  ) +
  theme_minimal()
```

## 9. Distribuția normală și probabilități

```r
# P(X ≤ LIMITA)
pnorm(LIMITA, mean = MEDIA, sd = DEVIATIA_STANDARD)
```

```r
# P(LIMITA_INFERIOARA ≤ X ≤ LIMITA_SUPERIOARA)
pnorm(LIMITA_SUPERIOARA, mean = MEDIA, sd = DEVIATIA_STANDARD) -
  pnorm(LIMITA_INFERIOARA, mean = MEDIA, sd = DEVIATIA_STANDARD)
```

```r
# Vizualizarea unei distribuții normale
VALORI_X <- seq(MINIM_X, MAXIM_X, length.out = 500)
DENSITATE <- dnorm(VALORI_X, mean = MEDIA, sd = DEVIATIA_STANDARD)

plot(
  VALORI_X, DENSITATE,
  type = "l",
  main = "TITLU_GRAFIC",
  xlab = "VALORI",
  ylab = "DENSITATE"
)
```

## 10. Covarianță și corelație

```r
cov(DATASET$VARIABILA_X, DATASET$VARIABILA_Y, use = "complete.obs")
cor(DATASET$VARIABILA_X, DATASET$VARIABILA_Y, use = "complete.obs")
```

```r
# Matrice de corelații
VARIABILE_NUMERICE <- DATASET |>
  select(VARIABILA_1, VARIABILA_2, VARIABILA_3)

cor(VARIABILE_NUMERICE, use = "complete.obs")
```

## 11. Rețetă completă de analiză

```r
library(readr)
library(dplyr)
library(ggplot2)

# 1. Import
DATASET <- read_csv("NUME_FISIER.csv")

# 2. Verificare
head(DATASET)
str(DATASET)
summary(DATASET)

# 3. Curățare
DATE_ANALIZA <- DATASET |>
  filter(!is.na(VARIABILA_X), !is.na(VARIABILA_Y))

# 4. Statistici descriptive
mean(DATE_ANALIZA$VARIABILA_X, na.rm = TRUE)
median(DATE_ANALIZA$VARIABILA_X, na.rm = TRUE)
sd(DATE_ANALIZA$VARIABILA_X, na.rm = TRUE)

# 5. Distribuție
ggplot(DATE_ANALIZA, aes(x = VARIABILA_X)) +
  geom_histogram(bins = NUMAR_INTERVALe) +
  labs(title = "TITLU", x = "AXA X", y = "Frecvență") +
  theme_minimal()

# 6. Relație
ggplot(DATE_ANALIZA, aes(x = VARIABILA_X, y = VARIABILA_Y)) +
  geom_point() +
  labs(title = "TITLU", x = "AXA X", y = "AXA Y") +
  theme_minimal()

cor(DATE_ANALIZA$VARIABILA_X, DATE_ANALIZA$VARIABILA_Y,
    use = "complete.obs")
```
