## 4. Tipurile variabilelor

Tipul variabilei determină ce operații și statistici au sens.


| Tipul variabilei | Cum apare | Exemplu AirBox |
|---|---|---|
| Numerică discretă | Se numără | numărul persoanelor |
| Numerică continuă | Se măsoară | temperatura, CO₂ |
| Categorică nominală | Categorii fără ordine | camera |
| Categorică ordinală | Categorii cu ordine | bună / moderată / slabă |

**Numerică discretă**
Ia valori separate, de obicei numere întregi. De exemplu, într-o cameră pot fi 2.0 sau 4.0 persoane, dar nu 2.4 persoane.


**Numerică continuă**
Poate lua, teoretic, orice valoare dintr-un interval. Temperatura rămâne conceptual continuă chiar dacă senzorul afișează un număr finit de zecimale.


**Categorică nominală**
Grupează observațiile în categorii care nu au o ordine naturală: laborator, clasă, exterior.


**Categorică ordinală**
Grupează observațiile în categorii ordonate: calitate bună, moderată sau slabă.

```r
# Ilustratie vizuala
library(ggplot2)

# Folosim coloanele reale și două categorii derivate din ele
hour <- as.integer(format(airbox$time, "%H"))
temperature <- airbox$temperature
sun_status <- factor(
  airbox$sun,
  levels = c(0, 1),
  labels = c("Fără soare", "Soare")
)
pm10_level <- cut(
  airbox$pm10,
  breaks = c(-Inf, 3, 4, Inf),
  labels = c("scăzut", "mediu", "ridicat"),
  ordered_result = TRUE
)

date_tipuri <- rbind(
  data.frame(
    tip = "Numerică discretă: ora",
    categorie = names(table(hour)),
    frecventa = as.numeric(table(hour))
  ),
  data.frame(
    tip = "Numerică continuă: temperature",
    categorie = names(table(cut(temperature, breaks = 4))),
    frecventa = as.numeric(table(cut(temperature, breaks = 4)))
  ),
  data.frame(
    tip = "Categorică nominală: sun",
    categorie = names(table(sun_status)),
    frecventa = as.numeric(table(sun_status))
  ),
  data.frame(
    tip = "Categorică ordinală: nivel PM10",
    categorie = names(table(pm10_level)),
    frecventa = as.numeric(table(pm10_level))
  )
)

ggplot(date_tipuri, aes(x = categorie, y = frecventa)) +
  geom_col(fill = "#4292C6") +
  facet_wrap(~tip, scales = "free_x", ncol = 2) +
  labs(x = NULL, y = "Frecvență") +
  theme_minimal() +
  theme(axis.text.x = element_text(angle = 30, hjust = 1))
```

**Variabile string (`character`)**
Un **string** este o secvență de caractere scrisă între ghilimele, de exemplu `"Laborator"`, `"Senzor_01"` sau `"750"`. În R, vectorii care conțin text au tipul `character`. Ei pot stoca denumiri, identificatori, răspunsuri sau observații. Chiar dacă un string conține cifre, R îl tratează ca text și nu poate face direct calcule cu el.

„String” și „variabilă categorială” nu înseamnă exact același lucru: `character` descrie **cum este stocată** informația în R, iar „categorială” descrie **rolul statistic** al variabilei. De exemplu, valorile `"laborator"`, `"clasă"` și `"exterior"` pot fi păstrate inițial ca text și apoi convertite într-un `factor` pentru a fi tratate explicit drept categorii.

### 4.1. Cum verificăm și convertim tipul variabilelor


`class()` ne arată tipul curent al unei variabile. Funcțiile `as.numeric()`, `as.factor()` și `as.character()` creează versiuni convertite; pentru a păstra conversia într-un data frame, atribuim rezultatul înapoi coloanei cu `<-`.

```r
# Copiem trei observații și simulăm tipuri importate greșit
airbox_tipuri <- head(airbox, 3)
airbox_tipuri$co2 <- paste(airbox_tipuri$co2, "ppm")
airbox_tipuri$sun <- ifelse(airbox_tipuri$sun == 1, "soare", "fără soare")
airbox_tipuri$time <- format(airbox_tipuri$time)

# Verificăm tipurile inițiale
class(airbox_tipuri$co2)
class(airbox_tipuri$sun)
class(airbox_tipuri$time)

# Convertim și salvăm tipurile potrivite
airbox_tipuri$co2 <- readr::parse_number(airbox_tipuri$co2)
airbox_tipuri$sun <- as.factor(airbox_tipuri$sun)
airbox_tipuri$time <- as.character(airbox_tipuri$time)

# Verificăm toate coloanele după conversie
str(airbox_tipuri)
```

**Atenție la conversii:**
- `as.numeric("620")` produce numărul `620`, dar `as.numeric("620 ppm")` produce `NA`, deoarece textul conține și litere. Pentru asemenea valori putem folosi `readr::parse_number("620 ppm")`.
- `as.numeric()` aplicat direct unui `factor` poate returna codurile interne ale categoriilor, nu valorile afișate. Pentru un factor care conține numere scrise ca etichete, folosim `as.numeric(as.character(variabila))`.

```r
# Conversia unui text construit din prima valoare CO2
co2_text <- paste(airbox$co2[1], "ppm")
readr::parse_number(co2_text)

# Conversia sigură factor -> numeric
co2_factor <- factor(as.character(airbox$co2[1:3]))
as.numeric(as.character(co2_factor))
```

## 5. Valorile lipsă

### 5.1. Ce este o valoare lipsă?


- senzorul nu a transmis date;
- dispozitivul s-a oprit;
- măsurarea a fost invalidă;
- fișierul a fost importat greșit.


**Valorile lipsă nu sunt 0**


Eliminarea valorilor lipsă este o decizie analitică, nu doar un truc de cod. Investigăm cauza înainte să eliminăm valorile.

În R, o valoare lipsă este reprezentată prin `NA` (*Not Available*).

```r
temperatures <- airbox$temperature[1:4]
temperatures[3] <- NA
```

Valorile lipsă pot afecta în mod negativ interpretările statistice. De exemplu calculul mediei.

```r
mean(temperatures)
```

R nu poate calcula media completă fără să știe ce trebuie făcut cu valoarea lipsă. Argumentul `na.rm = TRUE` îi spune lui R să elimine valorile lipsă din calcul.

```r
mean(temperatures, na.rm = TRUE)
```

Funcția `is.na()` identifică pozițiile în care apar valori lipsă

```r
is.na(temperatures)
```

```r
# Ilustratie vizuala
library(ggplot2)

date_co2 <- airbox[1:8, c("time", "co2")]
date_co2$co2[c(4, 5)] <- NA
momente_lipsa <- date_co2[is.na(date_co2$co2), "time", drop = FALSE]

ggplot(date_co2, aes(x = time, y = co2)) +
  geom_line(linewidth = 0.8, na.rm = TRUE) +
  geom_point(size = 2, na.rm = TRUE) +
  geom_vline(
    data = momente_lipsa,
    aes(xintercept = time),
    linetype = "dotted",
    color = "red"
  ) +
  geom_text(
    data = momente_lipsa,
    aes(x = time, y = min(date_co2$co2, na.rm = TRUE) - 5, label = "NA"),
    color = "red"
  ) +
  labs(
    title = "NA produce o întrerupere, nu o valoare zero",
    x = "Moment",
    y = "CO2 (ppm)"
  ) +
  theme_minimal()
```

## 6. Alegerea graficului

Graficul se alege în funcție de întrebare și de tipul variabilelor.

## 8. Cum comprimăm sute de măsurători

Un tabel cu sute sau mii de rânduri nu poate fi înțeles doar prin privire. Statistica ne ajută să comprimăm informația fără să pierdem povestea datelor.


Întrebări importante:
- Care este valoarea tipică?
- Cât de răspândite sunt observațiile?
- Ce formă are distribuția?
- Există vârfuri sau valori extreme?

```r
# Ilustratie vizuala
library(ggplot2)

repere <- data.frame(
  statistica = c("Media", "Mediana"),
  valoare = c(mean(airbox$co2), median(airbox$co2))
)

ggplot(airbox, aes(x = co2)) +
  geom_histogram(bins = 6, fill = "grey85", color = "white") +
  geom_vline(
    data = repere,
    aes(xintercept = valoare, color = statistica, linetype = statistica),
    linewidth = 1.2
  ) +
  geom_rug() +
  scale_color_manual(values = c("Media" = "#D7301F", "Mediana" = "#2C7FB8")) +
  scale_linetype_manual(values = c("Media" = "solid", "Mediana" = "dashed")) +
  labs(
    title = "Un rezumat vizual al măsurătorilor",
    x = "CO2 (ppm)",
    y = "Frecvență",
    color = NULL,
    linetype = NULL
  ) +
  theme_minimal()
```

## 9. Cum sunt distribuite valorile?

**Distribuția** descrie valorile pe care le ia o variabilă și frecvența cu care apar acestea.

```r
ggplot(airbox, aes(x = co2)) +
  geom_histogram(bins = 20) +
  labs(
    title = "Distribuția valorilor CO₂",
    x = "CO₂ (ppm)",
    y = "Frecvență"
  )
```

De ce facem asta și nu ne uităm doar la medie?

Medie poate ascunde:
- valori stabile în jurul mediei;
- alternanțe mari între valori mici și mari;
- o distribuție asimetrică;
- câteva vârfuri extreme.


De aceea, privim împreună:
1. **forma** distribuției;
2. **centrul** distribuției;
3. **răspândirea** distribuției.

## 14. Distribuția normală


Distribuția normală:
- are formă de clopot;
- este simetrică;
- are media, mediana și modul în același punct;
- este definită de media $\mu$ și deviația standard $\sigma$.


Notația este:


$$
X\sim N(\mu,\sigma^2)
$$


- $\mu$ stabilește centrul;
- $\sigma$ stabilește răspândirea;
- $\sigma^2$ este varianța.


> Nu toate variabilele din natură sunt distribuite normal. Nivelurile de poluare pot fi asimetrice și pot avea vârfuri extreme.


### 14.1. Vizualizarea distribuției normale în R

```r
library(ggplot2)

mu_temperature <- mean(airbox$temperature)
sd_temperature <- sd(airbox$temperature)
x <- seq(
  mu_temperature - 4 * sd_temperature,
  mu_temperature + 4 * sd_temperature,
  length.out = 400
)
date_normala <- data.frame(
  x = x,
  densitate = dnorm(x, mean = mu_temperature, sd = sd_temperature)
)

ggplot(date_normala, aes(x = x, y = densitate)) +
  geom_line(linewidth = 1) +
  labs(
    title = "Distribuția normală pentru temperature",
    x = "Temperature (°C)",
    y = "Densitate"
  ) +
  theme_minimal()
```

## 15. Cum modifică $\mu$ și $\sigma$ distribuția normală?


- Schimbarea lui $\mu$ deplasează distribuția la stânga sau la dreapta.
- Schimbarea lui $\sigma$ modifică răspândirea distribuției.
- O valoare mică a lui $\sigma$ produce o curbă îngustă.
- O valoare mare a lui $\sigma$ produce o curbă mai lată.

```r
library(ggplot2)

mu_temperature <- mean(airbox$temperature)
sd_temperature <- sd(airbox$temperature)
x <- seq(
  mu_temperature - 4 * sd_temperature,
  mu_temperature + 4 * sd_temperature,
  length.out = 500
)
date_curbe <- rbind(
  data.frame(
    x = x,
    densitate = dnorm(x, mu_temperature - 3, sd_temperature),
    curba = "medie mai mică"
  ),
  data.frame(
    x = x,
    densitate = dnorm(x, mu_temperature, sd_temperature),
    curba = "media datelor"
  ),
  data.frame(
    x = x,
    densitate = dnorm(x, mu_temperature, sd_temperature * 1.5),
    curba = "deviație standard mai mare"
  )
)

ggplot(date_curbe, aes(x = x, y = densitate, color = curba)) +
  geom_line(linewidth = 1) +
  labs(
    title = "Efectul mediei și al deviației standard",
    x = "Temperature (°C)",
    y = "Densitate",
    color = NULL
  ) +
  theme_minimal()
```

> Deviația standard măsoară răspândirea valorilor, nu cât de „brusc” se schimbă seria în timp. Pentru caracterul brusc al schimbărilor trebuie să analizăm și ordinea temporală a observațiilor.

## 16. Cum interpretăm regula 68–95–99,7?


Pentru o distribuție aproximativ normală:
- aproximativ 68% dintre valori se află între $\mu-\sigma$ și $\mu+\sigma$;
- aproximativ 95% se află între $\mu-2\sigma$ și $\mu+2\sigma$;
- aproximativ 99,7% se află între $\mu-3\sigma$ și $\mu+3\sigma$.


Exemplu:


$$
\mu=22^\circ C,\qquad \sigma=1^\circ C
$$


| Interval | Proporție aproximativă |
|---|---:|
| 21–23 °C, adică $\mu\pm1\sigma$ | 68% |
| 20–24 °C, adică $\mu\pm2\sigma$ | 95% |
| 19–25 °C, adică $\mu\pm3\sigma$ | 99,7% |

```r
# Ilustratie vizuala
library(ggplot2)

mu <- mean(airbox$temperature)
sigma <- sd(airbox$temperature)
x <- seq(mu - 4 * sigma, mu + 4 * sigma, length.out = 800)
date_normala <- data.frame(x = x, densitate = dnorm(x, mu, sigma))

ggplot(date_normala, aes(x = x, y = densitate)) +
  geom_area(
    data = subset(date_normala, x >= mu - 3 * sigma & x <= mu + 3 * sigma),
    fill = "#BDD7E7"
  ) +
  geom_area(
    data = subset(date_normala, x >= mu - 2 * sigma & x <= mu + 2 * sigma),
    fill = "#6BAED6"
  ) +
  geom_area(
    data = subset(date_normala, x >= mu - sigma & x <= mu + sigma),
    fill = "#3182BD"
  ) +
  geom_line(linewidth = 0.9) +
  geom_vline(
    xintercept = mu + (-3:3) * sigma,
    linetype = "dotted",
    color = "grey40"
  ) +
  annotate("text", x = mu, y = max(date_normala$densitate) * 0.55, label = "68%") +
  annotate(
    "text",
    x = mu + 1.5 * sigma,
    y = max(date_normala$densitate) * 0.28,
    label = "95%"
  ) +
  annotate(
    "text",
    x = mu + 2.5 * sigma,
    y = max(date_normala$densitate) * 0.09,
    label = "99,7%"
  ) +
  labs(
    title = "Regula 68–95–99,7 pentru temperature",
    x = "Temperature (°C)",
    y = "Densitate"
  ) +
  theme_minimal()
```

> Regula nu trebuie aplicată automat datelor asimetrice sau unei distribuții care nu este aproximativ normală.

## 17. Cum calculăm probabilitatea pentru variabile continue?


### 17.1. Probability Density Function (PDF) — unde sunt concentrate valorile?


PDF arată unde sunt concentrate valorile unei variabile continue.


$$
P(a<X<b)=\text{aria de sub curbă între }a\text{ și }b
$$


Important:
- axa verticală reprezintă **densitatea**, nu probabilitatea directă;
- probabilitatea este o **arie**;
- aria totală de sub curbă este 1;
- înălțimea curbei într-un punct nu este probabilitatea exactă a acelui punct.


### 17.2. Cumulative Distribution Function (CDF) — câtă probabilitate s-a acumulat?


CDF arată câtă probabilitate s-a acumulat până la o anumită valoare:


$$
F(x)=P(X\leq x)
$$


Caracteristici:
- pornește de la 0;
- crește sau rămâne constantă;
- ajunge la 1.


### 17.3. PDF versus CDF


- **PDF:** unde sunt concentrate valorile?
- **CDF:** câtă probabilitate s-a acumulat până la $x$?


```r
# Ilustratie vizuala
library(ggplot2)

mu <- mean(airbox$temperature)
sigma <- sd(airbox$temperature)
x <- seq(mu - 4 * sigma, mu + 4 * sigma, length.out = 500)
date_pdf_cdf <- rbind(
  data.frame(
    x = x,
    valoare = dnorm(x, mu, sigma),
    functie = "PDF: densitate locală"
  ),
  data.frame(
    x = x,
    valoare = pnorm(x, mu, sigma),
    functie = "CDF: probabilitate acumulată"
  )
)

ggplot(date_pdf_cdf, aes(x = x, y = valoare, color = functie)) +
  geom_line(linewidth = 1) +
  facet_wrap(~functie, scales = "free_y", nrow = 1) +
  geom_hline(
    data = data.frame(
      functie = "CDF: probabilitate acumulată",
      prag = c(0, 1)
    ),
    aes(yintercept = prag),
    linetype = "dotted",
    color = "grey60",
    inherit.aes = FALSE
  ) +
  scale_color_manual(
    values = c(
      "PDF: densitate locală" = "#2C7FB8",
      "CDF: probabilitate acumulată" = "#D95F0E"
    ),
    guide = "none"
  ) +
  labs(x = "Temperature (°C)", y = "Valoare") +
  theme_minimal()
```

### 18.3. Probabilitatea unui interval

```r
limita_inferioara <- quantile(airbox$temperature, 0.25)
limita_superioara <- quantile(airbox$temperature, 0.75)

pnorm(
  limita_superioara,
  mean = mean(airbox$temperature),
  sd = sd(airbox$temperature)
) -
  pnorm(
    limita_inferioara,
    mean = mean(airbox$temperature),
    sd = sd(airbox$temperature)
  )
```

```r
# Ilustratie vizuala
library(ggplot2)

mu <- mean(airbox$temperature)
sigma <- sd(airbox$temperature)
limita_inferioara <- quantile(airbox$temperature, 0.25)
limita_superioara <- quantile(airbox$temperature, 0.75)
x <- seq(mu - 4 * sigma, mu + 4 * sigma, length.out = 600)
y <- dnorm(x, mean = mu, sd = sigma)
date_interval <- data.frame(
  x,
  y,
  in_interval = x >= limita_inferioara & x <= limita_superioara
)
probabilitate_interval <- pnorm(limita_superioara, mu, sigma) -
  pnorm(limita_inferioara, mu, sigma)

ggplot(date_interval, aes(x = x, y = y)) +
  geom_area(
    data = subset(date_interval, in_interval),
    fill = "#74C476"
  ) +
  geom_line(linewidth = 0.9) +
  annotate(
    "text",
    x = mu,
    y = max(y) * 0.45,
    label = paste0("Aria = ", round(probabilitate_interval, 3))
  ) +
  labs(
    title = "Probabilitatea dintre quartilele temperaturii",
    x = "Temperature (°C)",
    y = "Densitate"
  ) +
  theme_minimal()
```

Regula generală:


$$
P(a\leq X\leq b)=F(b)-F(a)
$$


> Probabilitatea unui interval este CDF la capătul drept minus CDF la capătul stâng.

### 21.2. Calculăm abaterile față de medie


$$
x_i-\bar{x}
$$


Pentru setul A:


| $x_i$ | $$x_i-\bar{x}$$|
|---:|---:|
| 21 | −2 |

| 23 | 0 |
| 24 | 1 |
| 25 | 2 |

Abaterile simple se anulează:


$$

### 21.3. Ridicăm abaterile la pătrat


$$
(x_i-\bar{x})^2
$$


Pătratele fac toate valorile nenegative și acordă o greutate mai mare abaterilor mari.


```r
# Ilustratie vizuala
library(ggplot2)

valori <- airbox$temperature[1:5]
media <- mean(valori)
date_abateri <- data.frame(
  valoare = valori,
  y = 1,
  abatere_patratica = (valori - media)^2
)

ggplot(date_abateri, aes(x = valoare, y = y)) +
  geom_point(size = 4) +
  geom_vline(xintercept = media, color = "#D7301F", linewidth = 1.2) +
  geom_segment(
    aes(xend = media, y = 1.05, yend = 1.05),
    color = "#3182BD",
    linewidth = 0.9
  ) +
  geom_text(
    aes(y = 1.18, label = paste0("abatere² = ", round(abatere_patratica, 2)))
  ) +
  coord_cartesian(ylim = c(0.75, 1.3)) +
  labs(title = "Distanțele față de medie", x = "Temperature (°C)", y = NULL) +
  theme_minimal() +
  theme(axis.text.y = element_blank(), axis.ticks.y = element_blank())
```

### 21.4. Calculăm media abaterilor pătratice

## 24. Cum măsoară covarianța mișcarea comună?


Covarianța verifică dacă abaterile a două variabile tind să aibă aceeași direcție.


$$
\operatorname{Cov}(X,Y)=
\frac{\sum_{i=1}^{n}(x_i-\bar{x})(y_i-\bar{y})}{n-1}
$$


Pentru fiecare observație, verificăm poziția celor două variabile față de mediile lor:
- dacă ambele sunt peste medie, produsul abaterilor este pozitiv;
- dacă ambele sunt sub medie, produsul abaterilor este pozitiv;
- dacă una este peste medie și cealaltă sub medie, produsul este negativ.


Interpretare:
- $\operatorname{Cov}(X,Y)>0$ — variabilele tind să se miște în aceeași direcție;
- $\operatorname{Cov}(X,Y)<0$ — variabilele tind să se miște în direcții opuse;
- $\operatorname{Cov}(X,Y)\approx0$ — nu apare o relație liniară evidentă.

```r
# Ilustratie vizuala
library(ggplot2)

x <- airbox$temperature
y <- airbox$humidity
produs_abatere <- (x - mean(x)) * (y - mean(y))
date_covarianta <- data.frame(
  temperature = x,
  humidity = y,
  semn_produs = ifelse(produs_abatere >= 0, "produs pozitiv", "produs negativ")
)

ggplot(
  date_covarianta,
  aes(x = temperature, y = humidity, color = semn_produs)
) +
  geom_point(size = 4) +
  geom_vline(xintercept = mean(x), linetype = "dashed", color = "grey40") +
  geom_hline(yintercept = mean(y), linetype = "dashed", color = "grey40") +
  scale_color_manual(
    values = c("produs pozitiv" = "#31A354", "produs negativ" = "#DE2D26")
  ) +
  labs(
    title = "Covarianța prin poziția față de medii",
    x = "Temperature (°C)",
    y = "Humidity (%)",
    color = NULL
  ) +
  theme_minimal()
```

```r
cov(
  airbox$temperature,
  airbox$humidity,
  use = "complete.obs"
)
```

> Semnul covarianței este ușor de interpretat. Magnitudinea este mai dificilă, deoarece depinde de unitățile de măsură.

## 25. Cum standardizăm relația prin corelație?


Covarianța are două limite practice:
- magnitudinea sa depinde de unitățile de măsură;
- unitatea sa este produsul unităților celor două variabile.


$$
[\operatorname{Cov}(X,Y)]=[\text{unitatea lui }X]\times[\text{unitatea lui }Y]
$$


De exemplu, covarianța dintre temperatură și CO₂ ar putea avea unitatea °C × ppm.


Corelația Pearson standardizează covarianța:


$$
r=\frac{\operatorname{Cov}(X,Y)}{s_Xs_Y}
$$


$$
-1\leq r\leq1
$$


Corelația:
- arată direcția relației liniare;
- descrie intensitatea relației liniare;
- nu are unitate de măsură;
- poate fi comparată între perechi de variabile măsurate în unități diferite.

```r
cov(
  airbox$temperature,
  airbox$humidity,
  use = "complete.obs"
)

cor(
  airbox$temperature,
  airbox$humidity,
  use = "complete.obs"
)
```

Dacă transformăm temperatura din °C în °F:
- covarianța se schimbă;
- corelația rămâne aceeași.


```r
# Ilustratie vizuala
library(ggplot2)

temperature_C <- airbox$temperature
temperature_F <- temperature_C * 9 / 5 + 32
co2 <- airbox$co2

eticheta_C <- paste0(
  "°C: cov = ", round(cov(temperature_C, co2), 1),
  "; r = ", round(cor(temperature_C, co2), 3)
)
eticheta_F <- paste0(
  "°F: cov = ", round(cov(temperature_F, co2), 1),
  "; r = ", round(cor(temperature_F, co2), 3)
)
date_unitati <- rbind(
  data.frame(temperature = temperature_C, co2, unitate = eticheta_C),
  data.frame(temperature = temperature_F, co2, unitate = eticheta_F)
)

ggplot(date_unitati, aes(x = temperature, y = co2)) +
  geom_point(size = 3, color = "#3182BD") +
  facet_wrap(~unitate, scales = "free_x", nrow = 1) +
  labs(x = "Temperature", y = "CO2 (ppm)") +
  theme_minimal()
```

## 26. Cum interpretăm direcția și intensitatea corelației?


Semnul lui $r$ indică direcția:
- $r>0$ — relație liniară pozitivă;
- $r<0$ — relație liniară negativă;
- $r\approx0$ — relație liniară slabă sau absentă.


Valoarea absolută $|r|$ indică intensitatea relației liniare:
- $|r|$ aproape de 1 — relație liniară puternică;
- $|r|$ aproape de 0 — relație liniară slabă.


Exemple vizuale:
- $r\approx+0{,}9$ — pozitivă și puternică;
- $r\approx+0{,}4$ — pozitivă și moderată;
- $r\approx0$ — fără o relație liniară evidentă;
- $r\approx-0{,}5$ — negativă și moderată;
- $r\approx-0{,}9$ — negativă și puternică.


```r
# Ilustratie vizuala
library(ggplot2)

relatii <- list(
  "temperature și battery" = airbox$battery,
  "temperature și humidity" = airbox$humidity,
  "temperature și pm10" = airbox$pm10,
  "temperature și co2" = airbox$co2
)
date_corelatii <- do.call(
  rbind,
  lapply(names(relatii), function(nume) {
    y <- relatii[[nume]]
    data.frame(
      temperature = airbox$temperature,
      y = y,
      relatie = paste0(nume, "\nr = ", round(cor(airbox$temperature, y), 2))
    )
  })
)

ggplot(date_corelatii, aes(x = temperature, y = y)) +
  geom_point(color = "#3182BD", size = 2) +
  facet_wrap(~relatie, scales = "free_y", ncol = 2) +
  labs(x = "Temperature (°C)", y = "Valoarea celeilalte variabile") +
  theme_minimal()
```

### 26.1. Vizualizarea corelației printr-un scatterplot

```r
ggplot(airbox, aes(x = temperature, y = humidity)) +
  geom_point() +
  labs(
    title = "Relația dintre temperatură și umiditate",
    x = "Temperatură (°C)",
    y = "Umiditate (%)"
  )
```

Într-un scatterplot urmărim:
- direcția relației;
- intensitatea relației;
- forma relației;
- observațiile extreme;
- eventualele grupuri distincte.


> Pearson măsoară relații liniare. Un tipar curbat poate avea o corelație apropiată de zero, chiar dacă între variabile există o relație clară.

## 27. Ce nu ne spune corelația?


Corelația este un indiciu, nu o explicație cauzală.


### 27.1. Corelația nu demonstrează cauzalitatea


O asociere nu arată automat ce variabilă o produce pe cealaltă.


### 27.2. Corelația Pearson măsoară numai relația liniară


Pearson poate rata un tipar curbat.


```r
# Ilustratie vizuala
library(ggplot2)

x <- as.numeric(scale(airbox$temperature))
y <- x^2
date_neliniara <- data.frame(
  temperature_standardizata = x,
  relatie_patratica = y
)

ggplot(
  date_neliniara,
  aes(x = temperature_standardizata, y = relatie_patratica)
) +
  geom_point(color = "#756BB1", size = 2.5) +
  geom_line(color = "#54278F", linewidth = 0.9) +
  labs(
    title = paste0("Relație clară, dar Pearson r = ", round(cor(x, y), 2)),
    x = "Temperature standardizată",
    y = "Temperature standardizată la pătrat"
  ) +
  theme_minimal()
```

### 27.3. Valorile extreme pot modifica rezultatul


Un singur punct neobișnuit poate modifica puternic coeficientul $r$.


```r
# Ilustratie vizuala
library(ggplot2)

temperature_initial <- airbox$temperature
humidity_initial <- airbox$humidity
temperature_cu_outlier <- c(temperature_initial, max(temperature_initial) + 8)
humidity_cu_outlier <- c(humidity_initial, max(humidity_initial) + 25)

eticheta_fara <- paste0(
  "Fără outlier: r = ",
  round(cor(temperature_initial, humidity_initial), 2)
)
eticheta_cu <- paste0(
  "Cu outlier: r = ",
  round(cor(temperature_cu_outlier, humidity_cu_outlier), 2)
)
date_outlier <- rbind(
  data.frame(
    temperature = temperature_initial,
    humidity = humidity_initial,
    grup = eticheta_fara,
    tip_punct = "observație"
  ),
  data.frame(
    temperature = temperature_cu_outlier,
    humidity = humidity_cu_outlier,
    grup = eticheta_cu,
    tip_punct = c(rep("observație", nrow(airbox)), "outlier")
  )
)

ggplot(
  date_outlier,
  aes(x = temperature, y = humidity, color = tip_punct)
) +
  geom_point(size = 3) +
  facet_wrap(~grup, nrow = 1) +
  scale_color_manual(values = c("observație" = "black", "outlier" = "red")) +
  labs(
    x = "Temperature (°C)",
    y = "Humidity (%)",
    color = NULL
  ) +
  theme_minimal()
```

### 27.4. O variabilă terță poate explica relația


Ventilația, numărul persoanelor sau activitatea din laborator pot influența simultan două măsurători.


```r
# Ilustratie vizuala
library(ggplot2)

date_variabila_terta <- data.frame(
  sun = factor(
    airbox$sun,
    levels = c(0, 1),
    labels = c("Fără soare", "Soare")
  ),
  temperature = airbox$temperature,
  humidity = airbox$humidity
)

ggplot(
  date_variabila_terta,
  aes(x = temperature, y = humidity, color = sun)
) +
  geom_point(size = 4) +
  scale_color_manual(
    values = c("Fără soare" = "#9ECAE1", "Soare" = "#084594")
  ) +
  labs(
    title = "Relația temperature–humidity, colorată după sun",
    x = "Temperature (°C)",
    y = "Humidity (%)",
    color = "Sun"
  ) +
  theme_minimal()
```

Exemplu:
> Temperatura și VOC pot crește împreună, dar aceasta nu demonstrează că temperatura cauzează creșterea VOC. Ambele pot fi influențate de numărul persoanelor, de ventilație sau de activitățile din laborator.


Formula corectă este:
> „Temperatura este asociată cu VOC.”


Nu:
> „Temperatura cauzează VOC.”
